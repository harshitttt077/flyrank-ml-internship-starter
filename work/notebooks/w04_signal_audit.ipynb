## 1. Hypotheses: which signals predict decay
We hypothesize that search presence consistency (`days_with_impressions`), striking distance SERP positions (11–25), and depressed CTR relative to impressions are strong multi-signal predictors of decay.


```python
import pandas as pd, numpy as np
df = pd.read_csv('../data/processed/refresh_feature_vector.csv')
corr = df[['is_declining_label', 'days_with_impressions', 'log_impressions_90d', 'avg_position', 'content_age_days', 'days_since_last_update', 'ctr', 'word_count']].corr()['is_declining_label'].sort_values(ascending=False)
print("Correlation with is_declining_label:")
print(corr)
```

## 2. Audit code & non-linear signal interactions
Examining decay rates across positioning tiers and freshness brackets.


```python
df['position_bucket'] = pd.cut(df['avg_position'], bins=[0, 3, 10, 20, 50, 100], labels=['Top 3', 'Page 1', 'Striking', 'Deep Page', 'Beyond'])
print("Decay Rate by Position Bucket:")
print(df.groupby('position_bucket', observed=False)['is_declining_label'].mean())
```

## 3. Findings & Key Predictors
Consistency of search impressions (`days_with_impressions`) and position volatility are vastly more predictive than raw content age.


```python
print("Signal Audit Key Takeaway: Multi-signal models must capture SERP positioning drift alongside engagement softness.")
```

## 4. Dead ends & surprises
Age alone has near-zero linear correlation with decay (+0.04), proving why simplistic 'older than 6 months' rules fail.
